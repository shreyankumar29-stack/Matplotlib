```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Reading Files
df=pd.read_csv('iris.csv')
df.sample(5)
```

```python
#Scatter Plot
plt.scatter(df['SepalLengthCm'], df['PetalLengthCm'])
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

```python
# Replaccing values
df['Species']= df['Species'].replace({'Iris-setosa':0, 'Iris-versicolor':1,'Iris-virginica':2})
df.sample(5)
```

```python
plt.scatter(df['SepalLengthCm'], df['PetalLengthCm'],c=df['Species'],cmap='jet',alpha=0.75)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
```

```python
# Plot Size
plt.figure(figsize=(15,7))
plt.scatter(df['SepalLengthCm'], df['PetalLengthCm'],c=df['Species'],cmap='jet',alpha=0.75)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
```
